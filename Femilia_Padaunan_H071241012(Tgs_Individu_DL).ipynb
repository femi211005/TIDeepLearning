# TUGAS 1: CNN (Convolutional Neural Network)
### Dataset: Fashion-MNIST

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

print(f"TensorFlow Version: {tf.ara_version if hasattr(tf, 'ara_version') else tf.__version__}")
```

```python
# 1. Load & Preprocess Dataset
(X_train, y_train), (X_test, y_test) = datasets.fashion_mnist.load_data()
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# 2. Membangun Arsitektur CNN
cnn_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

cnn_model.summary()
```

```python
# 3. Kompilasi & Training
cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

history_cnn = cnn_model.fit(X_train, y_train, epochs=10, batch_size=64,
                            validation_data=(X_test, y_test))
```

```python
# 4. Evaluasi
test_loss, test_acc = cnn_model.evaluate(X_test, y_test, verbose=2)
print(f"\nAkurasi CNN pada Data Testing: {test_acc * 100:.2f}%")
```

```python
# 5. Simpan Model
cnn_model.save('fashion_mnist_cnn_model.keras')
print("Model CNN berhasil disimpan!")
```

# TUGAS 2: RNN (Recurrent Neural Network / LSTM)
### Dataset: IMDb Reviews (Sentiment Analysis)

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.preprocessing import sequence
import matplotlib.pyplot as plt
```

```python
# 1. Load & Preprocess Dataset
max_features = 10000
maxlen = 200

print("Memuat dataset IMDb...")
(X_train, y_train), (X_test, y_test) = datasets.imdb.load_data(num_words=max_features)

X_train = sequence.pad_sequences(X_train, maxlen=maxlen)
X_test = sequence.pad_sequences(X_test, maxlen=maxlen)
```

```python
# 2. Membangun Arsitektur RNN (LSTM)
rnn_model = models.Sequential([
    layers.Embedding(max_features, 32, input_length=maxlen),
    layers.LSTM(64, dropout=0.2, recurrent_dropout=0.2),
    layers.Dense(1, activation='sigmoid')
])

rnn_model.summary()
```

```python
# 3. Kompilasi & Training
rnn_model.compile(optimizer='adam',
                  loss='binary_crossentropy',
                  metrics=['accuracy'])

history_rnn = rnn_model.fit(X_train, y_train, epochs=5, batch_size=64,
                            validation_data=(X_test, y_test))
```

```python
# 4. Evaluasi
test_loss, test_acc = rnn_model.evaluate(X_test, y_test, verbose=2)
print(f"\nAkurasi RNN pada Data Testing: {test_acc * 100:.2f}%")
```

```python
# 5. Simpan Model
rnn_model.save('imdb_rnn_model.keras')
print("Model RNN berhasil disimpan!")
```

## 📌 Kesimpulan

Berdasarkan implementasi dan pengujian yang telah dilakukan, berikut adalah kesimpulan dari kedua arsitektur Deep Learning yang digunakan:

**1. Convolutional Neural Network (CNN):**
- **Fokus Data:** Dirancang khusus untuk memproses data spasial, seperti citra atau gambar (menggunakan dataset Fashion-MNIST).
- **Mekanisme:** Menggunakan filter konvolusi dan pooling untuk mengekstrak fitur visual secara hierarkis (dari tepi, bentuk, hingga objek utuh).
- **Hasil:** Sangat efektif, cepat dilatih, dan menghasilkan tingkat akurasi yang tinggi (berkisar di angka 90%+) untuk tugas klasifikasi gambar.

**2. Recurrent Neural Network (RNN - LSTM):**
- **Fokus Data:** Dirancang khusus untuk memproses data sekuensial atau berurutan, seperti teks, kalimat, atau data deret waktu (*time-series* menggunakan dataset IMDb Reviews).
- **Mekanisme:** Dilengkapi dengan mekanisme memori (*hidden state/LSTM*) yang mampu mempertahankan konteks dan hubungan antar-kata dalam sebuah kalimat.
- **Hasil:** Sangat andal dalam memahami urutan data teks dan melakukan analisis sentimen secara akurat.

Melalui tugas ini, dapat disimpulkan bahwa pemilihan arsitektur Deep Learning sangat bergantung pada karakteristik jenis data yang dihadapi—di mana CNN optimal untuk data visual/gambar, sedangkan RNN optimal untuk data berurutan seperti teks.
